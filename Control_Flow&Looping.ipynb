```python
#for i in range(n):
for i in range(5):
    print(i)
range(5)        # 0 → 4
range(1, 5)     # 1 → 4
range(1, 10, 2) # 1,3,5,7,9

```

```python
#for val in arr:
arr = [10, 20, 30]

for val in arr:
    print(val)
#Use this when you need the index
for i in range(len(arr)):
    print("index:",arr[i])
```

```python
#while start < end:
start = 0
end = 5

while start < end:
    print(start)
    start += 1

"""⚠️ VERY IMPORTANT
❌ Forgetting to update condition → infinite loop"""
    
```

```python
#if, elif, else:
age = 20

if age < 18:
    print("Minor")
elif age < 65:
    print("Adult")
else:
    print("Senior")
```

```python
#break, continue
for i in range(10):
    if i == 5:
        break
    print("breaked:",i)

for i in range(5):
    if i == 2:
        continue
    print("contin:",i)
```

Putting EVERYTHING Together 🔥

🔍 Example: Find first even number

```python
arr = [3, 7, 9, 10, 15]

for val in arr:
    if val % 2 == 0:
        print("First even:", val)
        break
```

🔄 Example: Two-pointer while start < end

```python
arr = [1, 2, 3, 4, 5]
start = 0
end = len(arr) - 1

while start < end:
    print(arr[start], arr[end])
    start += 1
    end -= 1
```

| Situation           | Use                |
| ------------------- | ------------------ |
| Known repetitions   | `for range()`      |
| Loop through list   | `for val in arr`   |
| Unknown repetitions | `while`            |
| Make decisions      | `if / elif / else` |
| Exit loop early     | `break`            |
| Skip iteration      | `continue`         |
